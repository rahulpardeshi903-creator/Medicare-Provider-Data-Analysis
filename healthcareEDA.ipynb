```python
import pandas as pd

df = pd.read_csv("C:\\Users\\Admin\\Documents\\healthcare dataset\\Healthcare Providers.csv")

# df.info()

# print(df.head())

df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

print(df)
```

```python
df
```

```python
df['full_name_of_the_provider'] = df['first_name_of_the_provider'] + ' ' + df['last_name/organization_name_of_the_provider']
```

```python
df.drop(['first_name_of_the_provider', 'last_name/organization_name_of_the_provider'], axis = 1, inplace = True)
```

```python
df
```

```python
df['street_address_of_the_provider'] = df['street_address_1_of_the_provider'].combine_first(df['street_address_2_of_the_provider'])
```

```python
df
```

```python
df.drop(['street_address_1_of_the_provider', 'street_address_2_of_the_provider'], axis = 1, inplace = True)
```

```python
df
```

```python
null_counts = df.isnull().sum()
null_percentage = (df.isnull().sum() / len(df)) * 100
null_df = pd.DataFrame({'null_count': null_counts,
                        'null_percentage': null_percentage}).sort_values('null_percentage', ascending=False)
print(null_df)
```

```python
df['credentials_of_the_provider'].fillna('Unknown', inplace = True)
df['gender_of_the_provider'].fillna('Unknown', inplace = True)
```

```python
df
```

```python
df.duplicated().sum()
```

```python
df['gender_of_the_provider'] = df['gender_of_the_provider'].replace('Unknown' , 'U')
```

```python
df['number_of_services'] = pd.to_numeric(df['number_of_services'], errors = 'coerce')
df['number_of_medicare_beneficiaries'] = pd.to_numeric(df['number_of_medicare_beneficiaries'], errors = 'coerce')
df['number_of_distinct_medicare_beneficiary/per_day_services'] = pd.to_numeric(df['number_of_distinct_medicare_beneficiary/per_day_services'], errors = 'coerce')
df['average_medicare_allowed_amount'] = pd.to_numeric(df['average_medicare_allowed_amount'], errors = 'coerce')
df['average_submitted_charge_amount'] = pd.to_numeric(df['average_submitted_charge_amount'], errors = 'coerce')
df['average_medicare_payment_amount'] = pd.to_numeric(df['average_medicare_payment_amount'], errors = 'coerce')
df['average_medicare_standardized_amount'] = pd.to_numeric(df['average_medicare_standardized_amount'], errors = 'coerce')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(df['average_medicare_payment_amount'], bins = 30)
plt.show()
```

```python
top_specialities = df.groupby('provider_type')['average_medicare_payment_amount'].mean().sort_values(ascending = False).head(10)

top_specialities.plot(kind='bar')
plt.title('Top 10 Specialities by Average medicare payment')
plt.ylabel('Avrage Payment')
plt.show()
```

```python
top_countries = df.groupby('country_code_of_the_provider')['average_medicare_payment_amount'].mean().sort_values(ascending = False).head(10)

top_specialities.plot(kind='bar')
plt.title('country by Average medicare payment')
plt.ylabel('Avrage Payment')
plt.show()
```

```python
top_cities = df.groupby('city_of_the_provider')['average_medicare_payment_amount'].mean().sort_values(ascending = False).head(10)

top_specialities.plot(kind='bar')
plt.title('Top 10 city by Average medicare payment')
plt.ylabel('Avrage Payment')
plt.show()
```
